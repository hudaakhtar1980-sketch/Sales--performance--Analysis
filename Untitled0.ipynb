### Goal
Analyze sales performance across regions, categories and time to identify
best/worst performing products and seasonality.

### Tools Used
- Python
- Pandas
- NumPy
- Matplotlib
- Google Colab



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name, encoding="latin1")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print("COLUMN NAMES:")
print(df.columns.tolist())

print("\nDATASET INFORMATION:")
df.info()

print("\nMISSING VALUES:")
display(df.isnull().sum())

print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())
```

```python
print("BASIC STATISTICAL SUMMARY:")
display(df.describe())

print("\nMean Sales:", round(df["Sales"].mean(), 2))
print("Median Sales:", round(df["Sales"].median(), 2))
print("Mode Sales:", df["Sales"].mode().iloc[0])
print("Most Common Category:", df["Category"].mode().iloc[0])
print("Most Common Region:", df["Region"].mode().iloc[0])
```

```python
clean = df.copy()

clean["Order_Date"] = pd.to_datetime(
    clean["Order_Date"],
    dayfirst=True,
    errors="coerce"
)

clean["Ship_Date"] = pd.to_datetime(
    clean["Ship_Date"],
    dayfirst=True,
    errors="coerce"
)

clean["Sales"] = pd.to_numeric(
    clean["Sales"],
    errors="coerce"
)

clean = clean.drop_duplicates()

clean = clean.dropna(
    subset=[
        "Order_ID",
        "Order_Date",
        "Sales",
        "Region",
        "Category",
        "Product_Name"
    ]
)

print("Data cleaning completed!")
print("Rows after cleaning:", clean.shape[0])
print("Duplicate rows:", clean.duplicated().sum())
```

```python
clean["Year"] = clean["Order_Date"].dt.year
clean["Month"] = clean["Order_Date"].dt.month
clean["Month_Name"] = clean["Order_Date"].dt.month_name()

clean["Year_Month"] = (
    clean["Order_Date"]
    .dt.to_period("M")
    .astype(str)
)

clean["Quarter"] = (
    "Q" + clean["Order_Date"].dt.quarter.astype(str)
)

clean["Day_of_Week"] = clean["Order_Date"].dt.day_name()

clean["Shipping_Days"] = (
    clean["Ship_Date"] - clean["Order_Date"]
).dt.days

print("Feature engineering completed!")

display(clean.head())
```

```python
total_revenue = clean["Sales"].sum()

total_orders = clean["Order_ID"].nunique()

total_customers = clean["Customer_ID"].nunique()

total_products = clean["Product_ID"].nunique()

order_totals = clean.groupby("Order_ID")["Sales"].sum()

average_order_value = order_totals.mean()

print("========== SALES KPIs ==========")

print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Unique Products: {total_products:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")

print("\nNOTE:")
print("Profit Margin cannot be calculated because Profit/Cost is not available.")
print("Conversion Rate cannot be calculated because conversion data is not available.")
```

```python
yearly_sales = (
    clean.groupby("Year")["Sales"]
    .sum()
    .sort_index()
)

print("YEARLY SALES:")
display(yearly_sales)

plt.figure(figsize=(9, 5))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    clean.groupby("Year_Month")["Sales"]
    .sum()
    .sort_index()
)

display(monthly_sales)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
regional_sales = (
    clean.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("SALES BY REGION:")
display(regional_sales)

plt.figure(figsize=(9, 5))

regional_sales.plot(kind="bar")

plt.title("Sales Performance by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

best_region = regional_sales.idxmax()
worst_region = regional_sales.idxmin()

best_region_sales = regional_sales.max()
worst_region_sales = regional_sales.min()

print("Best Region:", best_region)
print("Best Region Sales:", round(best_region_sales, 2))

print("Lowest Performing Region:", worst_region)
print("Lowest Region Sales:", round(worst_region_sales, 2))
```

```python
category_sales = (
    clean.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("SALES BY CATEGORY:")
display(category_sales)

plt.figure(figsize=(9, 5))

category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()

best_category = category_sales.idxmax()
worst_category = category_sales.idxmin()

best_category_sales = category_sales.max()
worst_category_sales = category_sales.min()

print("Best Category:", best_category)
print("Lowest Performing Category:", worst_category)
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Share by Category")

plt.show()
```

```python
subcategory_sales = (
    clean.groupby("Sub_Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("SUB-CATEGORY SALES:")
display(subcategory_sales)

plt.figure(figsize=(12, 6))

subcategory_sales.plot(kind="bar")

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

best_subcategory = subcategory_sales.idxmax()
worst_subcategory = subcategory_sales.idxmin()

best_subcategory_sales = subcategory_sales.max()
worst_subcategory_sales = subcategory_sales.min()

print("Best Sub-Category:", best_subcategory)
print("Lowest Performing Sub-Category:", worst_subcategory)
```

```python
product_sales = (
    clean.groupby("Product_Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_10_products = product_sales.head(10)

print("TOP 10 PRODUCTS:")
display(top_10_products)

plt.figure(figsize=(12, 7))

top_10_products.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
bottom_10_products = (
    product_sales
    .sort_values()
    .head(10)
)

print("BOTTOM 10 PRODUCTS:")
display(bottom_10_products)

plt.figure(figsize=(12, 7))

bottom_10_products.sort_values(
    ascending=False
).plot(kind="barh")

plt.title("Bottom 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
state_sales = (
    clean.groupby("State")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_states = state_sales.head(10)

print("TOP 10 STATES:")
display(top_states)

plt.figure(figsize=(11, 6))

top_states.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 States by Sales")
plt.xlabel("Total Sales")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
segment_sales = (
    clean.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("CUSTOMER SEGMENT SALES:")
display(segment_sales)

plt.figure(figsize=(8, 5))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
order_region = (
    clean.groupby(
        ["Region", "Order_ID"]
    )["Sales"]
    .sum()
    .reset_index()
)

region_aov = (
    order_region.groupby("Region")["Sales"]
    .mean()
    .sort_values(ascending=False)
)

print("AVERAGE ORDER VALUE BY REGION:")
display(region_aov)

plt.figure(figsize=(9, 5))

region_aov.plot(kind="bar")

plt.title("Average Order Value by Region")
plt.xlabel("Region")
plt.ylabel("Average Order Value")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
quarterly_sales = (
    clean.groupby("Quarter")["Sales"]
    .sum()
    .reindex(["Q1", "Q2", "Q3", "Q4"])
)

print("QUARTERLY SALES:")
display(quarterly_sales)

plt.figure(figsize=(8, 5))

quarterly_sales.plot(kind="bar")

plt.title("Quarterly Sales Performance")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
shipping_analysis = (
    clean.groupby("Ship_Mode")
    .agg(
        Orders=("Order_ID", "nunique"),
        Sales=("Sales", "sum"),
        Average_Shipping_Days=("Shipping_Days", "mean")
    )
    .round(2)
    .sort_values("Sales", ascending=False)
)

print("SHIPPING MODE ANALYSIS:")
display(shipping_analysis)
```

```python
customer_sales = (
    clean.groupby(
        ["Customer_ID", "Customer_Name"]
    )["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("TOP 10 CUSTOMERS:")
display(customer_sales.head(10))
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    clean["Sales"],
    bins=40
)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FINAL SALES PERFORMANCE SUMMARY
# ============================================================

# Calculate all important values again
# so this cell does not depend on previous summary variables

total_revenue = clean["Sales"].sum()
total_orders = clean["Order_ID"].nunique()
total_customers = clean["Customer_ID"].nunique()

# Average Order Value
order_totals = clean.groupby("Order_ID")["Sales"].sum()
average_order_value = order_totals.mean()


# ---------------- REGIONAL PERFORMANCE ----------------

regional_sales_final = (
    clean.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

best_region = regional_sales_final.idxmax()
best_region_sales = regional_sales_final.max()

worst_region = regional_sales_final.idxmin()
worst_region_sales = regional_sales_final.min()


# ---------------- CATEGORY PERFORMANCE ----------------

category_sales_final = (
    clean.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

best_category = category_sales_final.idxmax()
best_category_sales = category_sales_final.max()

worst_category = category_sales_final.idxmin()
worst_category_sales = category_sales_final.min()


# ---------------- SUB-CATEGORY PERFORMANCE ----------------

subcategory_sales_final = (
    clean.groupby("Sub_Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

best_subcategory = subcategory_sales_final.idxmax()
best_subcategory_sales = subcategory_sales_final.max()

worst_subcategory = subcategory_sales_final.idxmin()
worst_subcategory_sales = subcategory_sales_final.min()


# ---------------- MONTHLY / SEASONAL PERFORMANCE ----------------

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

# Create Month_Name if it does not already exist
if "Month_Name" not in clean.columns:
    clean["Month_Name"] = clean["Order_Date"].dt.month_name()

seasonality_final = (
    clean.groupby("Month_Name")["Sales"]
    .sum()
    .reindex(month_order)
)

best_month = seasonality_final.idxmax()
best_month_sales = seasonality_final.max()

worst_month = seasonality_final.idxmin()
worst_month_sales = seasonality_final.min()


# ---------------- PRODUCT PERFORMANCE ----------------

product_sales_final = (
    clean.groupby("Product_Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_product = product_sales_final.index[0]
top_product_sales = product_sales_final.iloc[0]

lowest_product = product_sales_final.index[-1]
lowest_product_sales = product_sales_final.iloc[-1]


# ============================================================
# PRINT FINAL RESULTS
# ============================================================

print("=" * 70)
print("FINAL SALES PERFORMANCE SUMMARY")
print("=" * 70)

print(f"\nTotal Revenue: ${total_revenue:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")

print("\n--- REGIONAL PERFORMANCE ---")

print(
    f"Best Performing Region: {best_region} "
    f"(${best_region_sales:,.2f})"
)

print(
    f"Lowest Performing Region: {worst_region} "
    f"(${worst_region_sales:,.2f})"
)

print("\n--- CATEGORY PERFORMANCE ---")

print(
    f"Best Performing Category: {best_category} "
    f"(${best_category_sales:,.2f})"
)

print(
    f"Lowest Performing Category: {worst_category} "
    f"(${worst_category_sales:,.2f})"
)

print("\n--- SUB-CATEGORY PERFORMANCE ---")

print(
    f"Best Performing Sub-Category: {best_subcategory} "
    f"(${best_subcategory_sales:,.2f})"
)

print(
    f"Lowest Performing Sub-Category: {worst_subcategory} "
    f"(${worst_subcategory_sales:,.2f})"
)

print("\n--- SEASONAL PERFORMANCE ---")

print(
    f"Highest Sales Month: {best_month} "
    f"(${best_month_sales:,.2f})"
)

print(
    f"Lowest Sales Month: {worst_month} "
    f"(${worst_month_sales:,.2f})"
)

print("\n--- PRODUCT PERFORMANCE ---")

print(f"Top Selling Product: {top_product}")
print(f"Top Product Sales: ${top_product_sales:,.2f}")

print(f"\nLowest Selling Product: {lowest_product}")
print(f"Lowest Product Sales: ${lowest_product_sales:,.2f}")

print("\n--- KPI AVAILABILITY ---")

print(
    "Profit Margin: Not calculated "
    "(Profit/Cost data is not available in the dataset)"
)

print(
    "Conversion Rate: Not calculated "
    "(conversion data is not available in the dataset)"
)

print("\n" + "=" * 70)
print("FINAL SUMMARY GENERATED SUCCESSFULLY!")
print("=" * 70)
```
